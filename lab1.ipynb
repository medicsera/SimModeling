итерация - число - его квадрат

```python
def get_next(val):
    sq = val * val
    s = f"{sq:08d}"
    return int(s[2:6])

x = 2152
print(f"{'итер':<5} {'x':<8} {"x^2":<8} {'new_x':<8}")

for i in range(1, 101):
    x_prev = x
    x_sq = x**2
    x = get_next(x)
    print(f"{i:<5} 0.{x_prev:04d} 0.{x_sq:04d} 0.{x:04d}")
```
